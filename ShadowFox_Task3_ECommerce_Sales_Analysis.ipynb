## 4. Importing Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

#Set Visualization style
sns.set_style("whitegrid")

print("Libraries imported successfully")
```

##5. Loading the Dataset

```python
# Load the Superstore dataset

file_path=("/content/sample_-_superstore.xls")
df=pd.read_excel(file_path)

print("Dataset loaded successfully.")
print("Shape of the dataset:",df.shape)
```

## 6. Dataset Structure

The Superstore Sales dataset contains 10,194 records and 21 columns. Each record represents an order-related transaction containing information about customers, products, sales, discounts, profits, shipping, and geographical details.

The structure of the dataset will be examined to understand the available variables and their data types before beginning the detailed analysis.

```python
df.head()
```

## 7. Checking Column Names and Data Types

Understanding the column names and data types helps us identify the type of information available in the dataset and ensures that each variable is suitable for further analysis.

```python
# Display column names and their data types

print("Column Names:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
```

## 8. Checking Missing Values

Missing values can affect the accuracy of data analysis. Therefore, the dataset is checked to identify whether any columns contain missing or null values.

```python
# Check for missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

## 9. Checking Duplicate Records

Duplicate records can affect the accuracy of analysis by counting the same transaction more than once. Therefore, the dataset is checked for duplicate rows before proceeding with further analysis.

```python
# Check for duplicate records

duplicate_count=df.duplicated().sum()
print("Number of duplicate records:",duplicate_count)
```

## 10. Statistical Summary

A statistical summary is generated to understand the numerical variables in the dataset. It provides measures such as count, mean, standard deviation, minimum, maximum, and quartile values.

This helps us understand the general distribution and range of important numerical variables such as sales, quantity, discount, and profit.

```python
df.describe()
```

## 11. Converting Date Columns

The Order Date and Ship Date columns are converted into date format. This allows us to perform time-based analysis, such as studying sales and profit trends over different periods.

```python
# Convert date columns to datetime format

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Check the updated data types

print("Date columns converted successfully.\n")
print(df[['Order Date', 'Ship Date']].dtypes)
```

## 12. Creating Time-Based Features

To perform time-based analysis, additional features are created from the Order Date column. Year, Month, and Year-Month values will help us examine changes in sales and profit over time.

```python
# Create time-based features

df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Year-Month'] = df['Order Date'].dt.to_period('M').astype(str)

# Display the new columns

df[['Order Date', 'Year', 'Month', 'Year-Month']].head()
```

## 13. Sales and Profit Trends Over Time

Analyzing sales and profit over time helps identify changes in business performance across different periods. Monthly sales and profit will be calculated and visualized to observe overall trends.

```python
# Calculate monthly sales and profit

monthly_performance = df.groupby('Year-Month')[['Sales', 'Profit']].sum().reset_index()

monthly_performance
```

## 14. Monthly Sales and Profit Trend

```python
# Visualize monthly sales and profit

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_performance['Year-Month'],
    monthly_performance['Sales'],
    marker='o',
    label='Sales'
)

plt.plot(
    monthly_performance['Year-Month'],
    monthly_performance['Profit'],
    marker='o',
    label='Profit'
)

plt.title('Monthly Sales and Profit Trend')
plt.xlabel('Year-Month')
plt.ylabel('Amount')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

## 15. Interpretation of Sales and Profit Trends

The monthly analysis provides an overview of how sales and profit change across different periods. The visualization helps identify months with relatively higher or lower business performance and shows whether changes in sales are accompanied by similar changes in profit.

The sales values are generally higher than the profit values because profit represents the amount remaining after business costs are considered. The trend also helps identify periods where sales increase but profit does not increase at the same rate, which may indicate the influence of discounts, product mix, or other business factors.

Overall, the monthly trend provides a useful foundation for understanding changes in the company's sales and profitability over time.

## 16. Category-wise Sales and Profit Analysis

Product categories can have different levels of sales and profitability. In this section, sales and profit are grouped by category to compare the overall performance of each product category.

```python
# Calculate sales and profit by category

category_performance = (
    df.groupby('Category')[['Sales', 'Profit']]
    .sum()
    .sort_values('Sales', ascending=False)
)

category_performance
```

##17. Category-wise Sales and Profit Visualization

```python
# Visualize sales and profit by category

category_performance.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Sales and Profit by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Amount')
plt.xticks(rotation=0)
plt.legend(title='Metric')
plt.tight_layout()
plt.show()
```

## 18. Interpretation of Category-wise Sales and Profit

The category-wise analysis shows differences in sales and profitability across the product categories.

The comparison helps identify which categories contribute more to overall sales and which categories generate stronger or weaker profit. A category with high sales does not necessarily have the highest profit, because profitability can also be affected by discounts, product costs, and the type of products sold.

This analysis provides a useful understanding of the contribution of each product category to the overall business performance.

## 19. Sub-Category-wise Sales and Profit Analysis

Sub-categories provide a more detailed view of product performance. Analyzing sales and profit at the sub-category level helps identify products that contribute strongly to revenue and those that may require further attention because of lower profitability.

```python
# Calculate sales and profit by sub-category

subcategory_performance = (
    df.groupby('Sub-Category')[['Sales', 'Profit']]
    .sum()
    .sort_values('Sales', ascending=False)
)

subcategory_performance
```

## 20. Sub-Category-wise Sales and Profit Visualization

```python
# Visualize sales and profit by sub-category

subcategory_performance.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Sales and Profit by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Amount')
plt.xticks(rotation=45)
plt.legend(title='Metric')
plt.tight_layout()
plt.show()
```

## 21. Interpretation of Sub-Category Performance

The sub-category analysis provides a more detailed view of product performance within the different categories. The results show that sales and profit contributions vary across sub-categories.

Some sub-categories contribute strongly to sales as well as profit, while others show a difference between their sales contribution and profitability. This indicates that high sales volume does not always result in proportionally high profit.

The comparison helps identify strong-performing areas as well as sub-categories that may need further investigation based on their profitability and sales performance.

## 28. Discount vs Profit Analysis

Discounts can influence the profitability of products and orders. In this section, the relationship between discount levels and profit will be examined to understand whether changes in discount are associated with changes in profitability.

```python
# Calculate average profit for each discount level

discount_profit = (
    df.groupby('Discount')['Profit']
    .mean()
    .reset_index()
)

discount_profit
```

## 29. Discount vs Average Profit Visualization

```python
# Visualize the relationship between discount and average profit

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=discount_profit,
    x='Discount',
    y='Profit',
    marker='o'
)

plt.title('Discount vs Average Profit')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.tight_layout()
plt.show()
```

## 30. Interpretation of Discount vs Average Profit

The analysis shows how average profit changes across different discount levels. Comparing discount levels with average profit helps us understand the possible impact of discounts on profitability.

The visualization can be used to identify discount levels associated with relatively higher or lower average profit. Higher discounts may reduce the amount of profit earned on individual sales, although the overall effect can also depend on factors such as product category, sales volume, and product pricing.

Therefore, discount levels should be considered carefully when evaluating profitability.

## 31. Top and Bottom Performing Sub-Categories

Identifying the highest and lowest performing sub-categories helps us understand which product areas contribute strongly to sales and profitability. This analysis can also highlight sub-categories that may require further investigation.

```python
# Display top 5 and bottom 5 sub-categories based on profit

top_profit_subcategories = (
    df.groupby('Sub-Category')['Profit']
    .sum()
    .sort_values(ascending=False)
)

print("Top 5 Sub-Categories by Profit:")
display(top_profit_subcategories.head(5))

print("\nBottom 5 Sub-Categories by Profit:")
display(top_profit_subcategories.tail(5))
```

## Top and Bottom Performing Sub-Categories Visualization

```python
# Select top 5 and bottom 5 sub-categories based on total profit

selected_subcategories = pd.concat([
    top_profit_subcategories.head(5),
    top_profit_subcategories.tail(5)
])

# Visualize the selected sub-categories

plt.figure(figsize=(12, 6))

selected_subcategories.sort_values().plot(kind='bar')

plt.title('Top and Bottom Sub-Categories by Profit')
plt.xlabel('Sub-Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 33. Interpretation of Top and Bottom Performing Sub-Categories

The analysis highlights differences in profitability among the various sub-categories. The top-performing sub-categories contribute the highest total profit, while the bottom-performing sub-categories contribute relatively less profit or may show losses.

This comparison helps identify product areas that contribute positively to business profitability and areas that may require further investigation. Factors such as discounts, sales volume, product pricing, and operating costs may influence these differences.

The results can support further analysis and help in developing appropriate business strategies.

## 34. Correlation Analysis

Correlation analysis is used to understand the strength and direction of relationships between numerical variables. In this project, the relationships among Sales, Quantity, Discount, and Profit are examined.

A correlation value closer to +1 indicates a strong positive relationship, while a value closer to -1 indicates a strong negative relationship. A value close to 0 indicates a weak linear relationship.

```python
# Select important numerical variables

correlation_data = df[['Sales', 'Quantity', 'Discount', 'Profit']]

# Calculate the correlation matrix

correlation_matrix = correlation_data.corr()

correlation_matrix
```

## 35. Correlation Heatmap

A heatmap is used to visualize the correlation matrix. It makes it easier to identify positive and negative relationships between the numerical variables.

```python
# Create a correlation heatmap

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Sales, Quantity, Discount and Profit')
plt.tight_layout()
plt.show()
```

## 36. Interpretation of Correlation Analysis

The correlation analysis helps us understand the linear relationships among Sales, Quantity, Discount, and Profit.

The heatmap makes it easier to identify variables that move together positively or negatively. A positive correlation indicates that two variables tend to increase together, while a negative correlation indicates that one variable tends to decrease as the other increases.

The correlation values should be interpreted as relationships rather than proof of cause and effect. Other business factors may also influence sales and profitability.
