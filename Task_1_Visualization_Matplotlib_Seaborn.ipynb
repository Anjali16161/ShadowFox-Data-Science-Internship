## 1. Library Overview
### Matplotlib
Matplotlib is a Python library used to create static and customizable data visuALIZATIONS. It is useful for exploring data, presenting trends, and creating graphs for reports.

### Seaborn
Seaborn is a Python visualization library built on Matplotlib. It provides a simpler interface for creating attractive statistical graphics and works well pandas DataFrames.

### Why these Libraries?
Matplotlib provides detailed control over graphs, while Seaborn makes statistical visualizations easier to create. Together, they cover many common data visualization requirements.

## 2. Matplotlib - Common Graph Types
###2.1 Line Plot
A line plot connects data points to show changes or trends over a sequence.

**Use case:** Comparing monthly sales or tracking temperature over time.

### 2.2 Bar Chart
A bar chart uses rectangular bars to compare values across categories.

**Use case:** Comparing sales across different products.

### 2.3 Scatter Plot
A scatter plot displays individual points to show the relationship between two numerical variables.

**Use case:** Studying the relationship between study hours and marks.

### 2.4 Histogram
A histogram groups numerical values into intervals to show their distribution.

**Use case:** Understanding the distribution of students' marks.

### 2.5 Pie Chart
A pie chart represents parts of a whole using different slices.

**Use case:** Showing the percentage contribution of different categories.


```python
import matplotlib.pyplot as plt

months=["Jan","Feb","Mar","Apr","May"]
Sales=[120,150,135,180,210]

plt.plot(months,Sales,marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

###2.6 Bar Chart - Practical Example
A bar chart compares values across different categories using rectangular bars,

**Use case:** Comparing the sales performance of different products.

```python
products=["Laptop","Phone","Tablet","Watch"]
sales=[80,120,70,95]

plt.bar(products,sales)
plt.title("Product Sales Comparison")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

###2.7 Scatter Plot - Practical Example
A scatter plot displays individual data points to show the relationship between two numerical variables.

**Use case:** Examining whether study time is related to students' marks.


```python
study_hours=[1,2,3,4,5]
marks=[45,50,58,65,72]

plt.scatter(study_hours,marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

### 2.8 Histogram – Practical Example

A histogram shows how numerical values are distributed by grouping them into intervals.

**Use case:** Understanding the distribution of students' marks.

```python
marks=[45,52,58,61,65,67,70,72,75,8,80,82,85,88,92]
plt.hist(marks,bins=5)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

### 2.9 Pie Chart – Practical Example

A pie chart shows how different categories contribute to a complete total.

**Use case:** Showing the proportion of sales contributed by different products.

```python
products=["Laptop","Phone","Tablet","Watch"]
sales=[85,120,70,95]

plt.pie(sales,labels=products,autopct="%1.1f%%")
plt.title("Sales Contribution by Product")
plt.show()
```

## 3. Seaborn – Common Graph Types

Seaborn provides a high-level interface for creating statistical graphics. It works closely with pandas DataFrames and offers convenient functions for exploring relationships and distributions in data.

The following examples demonstrate some commonly used Seaborn plots.

```python
import seaborn as sns
import matplotlib.pyplot as plt

days=["Mon","Tue","Wed","Thrus","Fri"]
visitors=[120,145,135,160,175]

sns.lineplot(x=days,y=visitors,marker="o")
plt.title("Daily Visitors")
plt.xlabel("Day")
plt.ylabel("Visitors")
plt.show()
```

### 3.1 Bar Plot – Practical Example

A Seaborn bar plot compares a numerical value across different categories.

**Use case:** Comparing the average sales of different product categories.

```python
categories = ["Books", "Clothing", "Electronics", "Sports"]
average_sales = [42, 65, 88, 55]

sns.barplot(x=categories, y=average_sales)
plt.title("Average Sales by Category")
plt.xlabel("Category")
plt.ylabel("Average Sales")
plt.show()
```

### 3.2 Scatter Plot – Practical Example

A Seaborn scatter plot displays individual observations and helps identify relationships between numerical variables.

**Use case:** Examining the relationship between advertising spending and sales.

```python
ad_spend = [10, 15, 20, 25, 30, 35]
sales = [25, 32, 38, 45, 52, 60]

sns.scatterplot(x=ad_spend, y=sales)
plt.title("Advertising Spend vs Sales")
plt.xlabel("Advertising Spend")
plt.ylabel("Sales")
plt.show()
```

### 3.3 Histogram – Practical Example

A Seaborn histogram displays the distribution of numerical data by grouping values into intervals.

**Use case:** Analyzing the distribution of students' scores.

```python
scores = [42, 48, 51, 55, 58, 61, 64, 67, 70, 72, 75, 78, 82, 86, 90]

sns.histplot(scores, bins=5)
plt.title("Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

### 3.4 Box Plot – Practical Example

A box plot summarizes the distribution of numerical data and helps identify the spread and possible outliers.

**Use case:** Comparing the distribution of scores across different groups.

```python
groups = ["A", "A", "A", "A", "B", "B", "B", "B"]
scores = [65, 70, 72, 68, 75, 80, 78, 85]

sns.boxplot(x=groups, y=scores)
plt.title("Score Distribution by Group")
plt.xlabel("Group")
plt.ylabel("Score")
plt.show()
```

### 3.5 Heatmap – Practical Example

A heatmap represents numerical values using different levels of shading, making patterns easier to identify.

**Use case:** Visualizing correlations between different numerical variables.

```python
import pandas as pd

data = pd.DataFrame({
    "Math": [80, 75, 90, 65],
    "Science": [85, 70, 88, 72],
    "English": [78, 82, 85, 68]
})

correlation = data.corr()

sns.heatmap(correlation, annot=True)
plt.title("Subject Score Correlation")
plt.show()
```

### 3.6 Other Useful Seaborn Plots

Seaborn also provides several other visualization functions, including violin plots, count plots, pair plots, and regression plots. These are useful for statistical analysis and exploring relationships within datasets.

## 4. Matplotlib vs Seaborn – Comparison

| Feature | Matplotlib | Seaborn |
|---|---|---|
| Ease of use | Requires more code and customization | Easier for statistical plots |
| Customization | Very high | High, with Matplotlib support |
| Interactivity | Mainly static plots | Mainly static plots |
| Statistical visualization | Can be done manually | Strong built-in support |
| DataFrame support | Works with DataFrames | Designed to work well with DataFrames |
| Large datasets | Can handle large datasets, depending on the plot | Can become slower with complex statistical plots |
| Best suited for | Detailed and highly customized charts | Quick and attractive statistical visualizations |

## 5. Strengths and Weaknesses

### Matplotlib

**Strengths**
- Highly customizable.
- Suitable for many types of charts.
- Gives detailed control over plot elements.
- Works well for publication-quality static figures.

**Weaknesses**
- Can require more code for advanced visualizations.
- Statistical plots may need additional programming.

### Seaborn

**Strengths**
- Easy to use for statistical visualizations.
- Works naturally with pandas DataFrames.
- Provides attractive default styles.
- Requires less code for many common statistical plots.

**Weaknesses**
- Offers less low-level control than Matplotlib.
- Complex or highly customized visualizations may require Matplotlib.
- Very large datasets can make statistical plots slower.
